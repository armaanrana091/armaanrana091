```python
# 1. Create a List with fruit names
my_list = ['apple', 'watermelon', 'banana', 'cherry', 'papaya']
print("Original List:", my_list)

# 2. Access Items in the List
print("Accessing Item at Index 2:", my_list[2])
print("Accessing Item at Index 0:", my_list[0])

# 3. Update List (Add Items)
my_list.append('mango')
print("After appending 'mango':", my_list)

# Add a fruit ('orange') at a specific index
my_list.insert(3, 'orange')
print("After inserting 'orange' at index 3:", my_list)

# 4. Update List (Remove Items)
my_list.remove('banana')
print("After removing 'banana':", my_list)

# Remove item by index (removes item at index 1)
removed_item = my_list.pop(1)
print("After popping item at index 1:", my_list)
print("Removed Item:", removed_item)

# 5. Delete List
del my_list
```

```python
# Step 1: Create a list of tuples (initial records)
employees = [
    (1, 'John Doe', 5000.0, 'HR'),
    (2, 'Jane Smith', 6000.0, 'Finance'),
    (3, 'Emily Davis', 7000.0, 'IT')
]

# Step 2: Create (Insert) a new tuple
new_employee = (4, 'Michael Brown', 8000.0, 'Marketing')
employees.append(new_employee)

# Display all employees after insertion
print("Employees after insertion:")
for employee in employees:
    print(employee)

# Step 3: Access (Read) a tuple (e.g., Employee with ID 2)
def access_employee(employee_id):
    for employee in employees:
        if employee[0] == employee_id:
            return employee
    return None

# Accessing employee with ID 2
employee = access_employee(2)
print("\nAccessed Employee with ID 2:", employee)

# Step 4: Update a tuple (e.g., update salary of employee with ID 3)
def update_employee(employee_id, new_salary):
    for i, employee in enumerate(employees):
        if employee[0] == employee_id:
            # Update the tuple by creating a new tuple with the modified value
            employees[i] = (employee[0], employee[1], new_salary, employee[3])

# Updating the salary of employee with ID 3
update_employee(3, 7500.0)

# Display employees after update
print("\nEmployees after salary update:")
for employee in employees:
    print(employee)

# Step 5: Delete a tuple (e.g., delete employee with ID 1)
def delete_employee(employee_id):
    global employees
    employees = [employee for employee in employees if employee[0] != employee_id]

# Deleting employee with ID 1
delete_employee(1)

# Display employees after deletion
print("\nEmployees after deletion:")
for employee in employees:
    print(employee)
```
